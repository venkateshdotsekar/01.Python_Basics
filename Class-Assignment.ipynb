```python
class multipleFunctions:
    @staticmethod
    def oddEven():
        num = int(input("Enter the number: "))
        if num % 2 == 1:
            print("Odd Number")
            message = "Odd Number"
        else:
            print("Even Number")
            message = "Even Number"
        return message

    @staticmethod
    def BMI():
        BMI = float(input("Enter the BMI Index: "))
        if BMI < 18.5:
            print("Underweight")
            message = "Underweight"
        elif BMI < 24.9:
            print("Normal")
            message = "Normal"
        elif BMI < 29.9:
            print("Overweight")
            message = "Overweight"
        else:
            print("Very Overweight")
            message = "Very Overweight"
        return message

if __name__ == "__main__":
    print("Choose function to run:")
    print("1. Odd or Even")
    print("2. BMI Category")
    choice = input("Enter choice (1 or 2): ")
    if choice == '1':
        multipleFunctions.oddEven()
    elif choice == '2':
        multipleFunctions.BMI()
    else:
        print("Invalid choice.")
```
