```python
# print Odd Numbers in the list
```

```python
# Sample list of numbers
numbers = [20,10,16,19,25,1,276,188]

# Print odd numbers from the list
for num in numbers:
    if num % 2 != 0:
        print(num, "is odd")
```

```python
# print Even numbers in the list
```

```python
# Sample list of numbers
numbers = [20,10,16,19,25,1,276,188]

# Print even numbers from the list
for num in numbers:
    if num % 2 == 0:
        print(num, "is even")
```
