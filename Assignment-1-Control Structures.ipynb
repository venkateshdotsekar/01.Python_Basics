```python
#Body Mass Index
"""
<18.5 | Underweight
18.5-24.9 | Normal
25-29.9 | Overweight
>30 | Obese
"""
```

```python
weight=float(input("Enter the your weight: "))
if weight <=18.5:
    print("Your BMI is UNDERWEIGHT")
elif weight >18.5 and weight<=24.9:
    print("Your BMI is NORMAL")
elif weight >=25 and weight<29.9:
    print("Your BMI is OVERWEIGHT")
else:
    print("Your BMI is OBESE")
```
