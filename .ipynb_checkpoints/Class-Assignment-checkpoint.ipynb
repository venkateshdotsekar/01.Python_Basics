```python
# Create a class and function, and list out the items in the list
```

```python
class SubfieldsInAI:
    def __init__(self):
        self.subfields = [
            "Machine Learning",
            "Neural Networks",
            "Vision",
            "Robotics",
            "Speech Processing",
            "Natural Language Processing"
        ]

    def Subfields(self):
        print("Sub-fields in AI are:")
        for subfield in self.subfields:
            print(subfield)
```

```python
obj = SubfieldsInAI()
obj.Subfields()
```

```python
# Create a function that checks whether the given number is Odd or Even
```

```python
class OddEven:
    def OddEven(self):
        num = int(input("Enter a number: "))
        if num % 2 == 0:
            print(f"{num} is Even number")
        else:
            print(f"{num} is Odd number")

# Create an object and call the function
OddEven().OddEven()
```

```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female
```

```python
class ElegiblityForMarriage:
    def Elegible(self):
        gender = input("Your Gender:")    # e.g., Male or Female
        age = int(input("Your Age:"))     # e.g., 20
        if (gender.lower() == 'male' and age >= 21) or (gender.lower() == 'female' and age >= 18):
            print("ELIGIBLE")
        else:
            print("NOT ELIGIBLE")
# Create an object and call the function
obj = ElegiblityForMarriage()
obj.Elegible()
```

```python
# calculate the percentage of your 10th mark
```

```python
class FindPercent:
    def percentage(self):
        # Get marks for 5 subjects
        marks = []
        for i in range(1, 6):
            mark = int(input(f"Subject{i}= "))
            marks.append(mark)
        # Calculate total and percentage
        total = sum(marks)
        percentage = (total / 500) * 100   # Assuming each subject is out of 100
        print("Total   : ", total)
        print("Percentage : ", percentage)

# Create an object and call the function
obj = FindPercent()
obj.percentage()
```

```python
#print area and perimeter of triangle using class and functions
```

```python
class Triangle:
    def triangle(self):
        # Area Calculation
        Height = int(input("Height:"))
        Breadth = int(input("Breadth:"))
        print("Area formula: (Height*Breadth)/2")
        area = (Height * Breadth) / 2
        print("Area of Triangle: ", area)
        
        # Perimeter Calculation
        Height1 = int(input("Height1:"))
        Height2 = int(input("Height2:"))
        Breadth = int(input("Breadth:"))
        print("Perimeter formula: Height1+Height2+Breadth")
        perimeter = Height1 + Height2 + Breadth
        print("Perimeter of Triangle: ", perimeter)

# Object creation
triangle = Triangle()
triangle.triangle()
```
